Распределение Коши и распрдееление Гаусса с выборкой 50

```python
import numpy as np
import matplotlib.pyplot as plt


n = 50
np.random.seed(200)

sample_cauchy = np.random.standard_cauchy(size=n)
sample_gauss = np.random.normal(loc=0, scale=1, size=n)

bins_list = [5, 10, n]

fig, axs = plt.subplots(2, 3, figsize=(15, 8))

for i, bins in enumerate(bins_list):
    axs[0, i].hist(sample_cauchy, bins=bins, color='skyblue', edgecolor='black')
    axs[0, i].set_title(f"Коши, bins={bins}")
    axs[0, i].set_xlim(-10, 10)

    axs[1, i].hist(sample_gauss, bins=bins, color='lightgreen', edgecolor='black')
    axs[1, i].set_title(f"Гаусс, bins={bins}")
    axs[1, i].set_xlim(-5, 5)


plt.tight_layout()
plt.show()

```


<h3 align="left"> Центра́льные преде́льные теоре́мы (ЦПТ) — класс теорем в теории вероятностей, утверждающих, что сумма достаточно большого количества слабо зависимых случайных величин, имеющих примерно одинаковые масштабы (ни одно из слагаемых не доминирует, не вносит в сумму определяющего вклада), имеет распределение, близкое к нормальному.</h3>

```python
N = 10000

means_cauchy = [np.mean(np.random.standard_cauchy(size=n)) for _ in range(N)]

means_gauss = [np.mean(np.random.normal(loc=0, scale=1, size=n)) for _ in range(N)]

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.hist(means_cauchy, bins=100, color='lightcoral', edgecolor='black')
plt.title("Распределение Коши")
plt.xlim(-250, 250)

plt.subplot(1,2,2)
plt.hist(means_gauss, bins=100, color='lightseagreen', edgecolor='black')
plt.title("Распределение Гаусс")
plt.xlim(-1, 1)

plt.tight_layout()
plt.show()
```



<h3 align="left">для распределения Коши центральная предельная теорема не выполняется — средние не стремятся к нормальному распределению и не стабилизируются при увеличении числа выборок </h3>

```python
print("Мин:", np.min(means_cauchy))
print("Макс:", np.max(means_cauchy))
```

```python
print("Мин:", np.min(means_gauss))
print("Макс:", np.max(means_gauss))
```
