```python
import numpy as np

n = 4

np.random.seed(100)
A = np.random.randint(0, 100, (n, n))
print(A)
```

```python
min_element = max_element = A[0][0]
min_row = min_col = max_row = max_col = 0

for i in range(n):
    for j in range(n):
        if A[i][j] < min_element:
            min_element = A[i][j]
            min_row, min_col = i, j
        if A[i][j] > max_element:
            max_element = A[i][j]
            max_row, max_col = i, j

print(min_element)
print(min_row, min_col)
print(max_element)
print(max_row, max_col)

temp = A[min_row][min_col]
A[min_row][min_col] = A[max_row][max_col]
A[max_row][max_col] = temp

print(A)
```

```python
def mult(A, B):
    if A.shape[1] != B.shape[0]:
        raise ValueError("Нельзя умножить матрицы")

    rows_A, cols_A = A.shape
    cols_B = B.shape[1]

    C = np.zeros((rows_A, cols_B), dtype=A.dtype)

    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                C[i, j] += A[i, k] * B[k, j]

    return C

try:
    B = mult(A, A)
    B = mult(B, A)
    B = mult(B, A)

except ValueError as e:
    print(e)

print(B)
```

```python
np.random.seed(100)
A = np.random.randint(0, 100, size=(n, n))
print("Исходная матрица A:")
print(A)

# Находим и меняем местами min и max
min_index = np.unravel_index(np.argmin(A), A.shape)
max_index = np.unravel_index(np.argmax(A), A.shape)
print("Индексы мин/макс:")
print(min_index)
print(max_index)
print("Значения мин/макс:")
print(A[min_index])
print(A[max_index])

print(f"\nДо замены:")
print(f"\nA[{min_index}] = {A[min_index]}, A[{max_index}] = {A[max_index]}")

A[min_index], A[max_index] = A[max_index], A[min_index]

print(f"\nПосле замены:")
print(f"A[{min_index}] = {A[min_index]}, A[{max_index}] = {A[max_index]}")
print(A)

# Вычисляем A@A@A@A
B = np.dot(A, A)
B = np.dot(B, A)
B = np.dot(B, A)

print(f"\nA@A@A@A:")
#print(B)

C = A*A
C = C*A
C = C*A
print(f"\nA·A·A·A:")
print(C)
```

<h3 align="left">

<h3 align="left">
Параметры: математическое ожидание, дисперсия: 2, 10</h3>
<h3 align="left">
Параметр массива: Сумма элементов </h3>


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(100)

mu = 2          # математическое ожидание
sigma = np.sqrt(10)  # стандартное отклонение
n = 20

Arr = np.random.normal(mu, sigma, n) # Вид распределения: Гаусса

print("Массив из 20 элементов:")
print(Arr)
print(f"Сумма элементов: {np.sum(Arr):.4f}")
```

```python
# Гистограмма
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)

minval = min(Arr)
maxval = max(Arr)
bins = (maxval - minval) / 10


plt.hist(Arr, bins=np.arange(minval, maxval + bins, bins), color='skyblue', edgecolor='black')
plt.xlabel('Значения')
plt.ylabel('Частота')
plt.title('Гистограмма нормального распределения')

plt.grid(True, alpha=0.3)

# boxplot
plt.subplot(1, 2, 2)
plt.boxplot(Arr, vert=True, patch_artist=True)
plt.ylabel('Значения')
plt.title('Boxplot массива')
plt.grid(True, alpha=0.3)

# Добавляем точки на boxplot
plt.scatter(np.ones_like(Arr), Arr, alpha=0.6, color='red', zorder=3)

plt.tight_layout()
plt.show()
```
